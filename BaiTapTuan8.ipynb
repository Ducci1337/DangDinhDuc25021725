```python
'''W7A16 
Cho một mảng số nguyên không âm. Gọi 
mảng đó là "mảng đẹp" nếu mỗi phần tử 
x trong mảng xuất hiện đúng x lần. 
Kiểm tra xem mảng có phải là mảng đẹp 
hay không. Nếu không hãy xóa đi y số (y 
là nhỏ nhất có thể) để mảng còn lại thành 
mảng đẹp. '''
l=list(map(int,input().split()))
d={}
for x in l:
    if x in d:
        d[x]+=1
    else:
        d[x]=1
res=0
for x in d:
    if d[x]<x:
        res+=d[x]
    elif d[x]>x:
        res+=d[x]-x 
if res==0:
    print("YES")
else:
    print("NO",res)
```

```python
'''W7A17 
Tìm phần tử xuất hiện nhiều hơn n/2 lần 
trong mảng (nếu có) với độ phức tạp thời 
gian là O(n), độ phức tạp bộ nhớ là O(1), - Boyer–Moore. Nếu không có phần tử 
nào thỏa mãn thì không in gì '''
n=int(input())
l=list(map(int,input().split()))
count=0
candidate=None
for x in l:
    if count==0:
        candidate=x
    if x==candidate:
        count+=1
    else:
        count-=1
if l.count(candidate)>n//2:
    print(candidate)
```

```python
'''W7A18 
Sắp xếp có điều kiện: Số chẵn đứng 
trước, số lẻ đứng sau, chẵn tăng dần, lẻ 
giảm dần '''
n=int(input())
l=list(map(int,input().split()))    
even=[]
odd=[]
for x in l:
    if x%2==0:
        even.append(x)
    else:
        odd.append(x)
even.sort()
odd.sort(reverse=True)
print(even+odd)
```

```python
'''W7A19 
Merge 2 mảng đã sắp xếp (có thể tăng 
dần, có thể giảm dần) mà không dùng 
bất cứ thuật toán sắp xếp nào thành một 
mảng cũng có cùng thứ tự sắp xếp như 2 
mảng ban đầu'''
l1=list(map(int,input().split()))
l2=list(map(int,input().split()))
n1=len(l1)
n2=len(l2) 
incr1=True
incr2=True  
if n1>1 and l1[0]>l1[1]:
    incr1=False
if n2>1 and l2[0]>l2[1]:
    incr2=False
res=[]
i=j=0   

while i<n1 and j<n2:
    if incr1 and incr2:
        if l1[i]<l2[j]:
            res.append(l1[i])
            i+=1
        else:
            res.append(l2[j])
            j+=1
    elif not incr1 and not incr2:
        if l1[i]>l2[j]:
            res.append(l1[i])
            i+=1
        else:
            res.append(l2[j])
            j+=1
    elif incr1 and not incr2:
        if l1[i]<l2[j]:
            res.append(l1[i])
            i+=1
        else:
            res.append(l2[j])
            j+=1
    else:
        if l1[i]>l2[j]:
            res.append(l1[i])
            i+=1
        else:
            res.append(l2[j])
            j+=1
while i<n1:
    res.append(l1[i])
    i+=1
while j<n2:
    res.append(l2[j])
    j+=1
for i in res:
    print(i,end=' ')
  
```

```python
'''W7A20 
Cho dãy N số nguyên dương, tính tổng 
các số có k chữ số. Dòng đầu ghi N số 
nguyên dương, dòng sau ghi số nguyên 
dương k'''
l=list(map(int,input().split()))
k=int(input())
res=0
for x in l:
    if len(str(x))==k:
        res+=x  
print(res)
```

```python
'''W7A22 
Cho N tên sinh viên gồm cả họ và tên, 
tìm xem tên nào hay gặp nhất. Nếu có 
nhiều tên cùng số lần gặp thì lấy tên có 
thứ tự từ điển nhỏ nhất '''
n=int(input())
d={}    
for _ in range(n):
    name=input().strip().split()[-1]
    if name in d:
        d[name]+=1
    else:
        d[name]=1
max_count=max(d.values())
candidates=[name for name, count in d.items() if count==max_count]
candidates.sort()
print(candidates[0])
```

```python
'''W7A23 
Đếm số lượng tên trong xâu kí tự. Tên 
được định nghĩa là xâu có độ dài >=2 và 
có chữ cái đầu viết hoa, các chữ cái khác 
viết thường, chỉ có kí tự cái. '''
s=input().strip().split()
d={}
for word in s:
    if len(word) >= 2 and word[0].isupper() and word[1:].islower() and word.isalpha():
        d[word] = d.get(word, 0) + 1
print(len(d))
```

```python
'''W7A24 
Cho xâu kí tự, xóa các từ giống nhau liền 
kề, chỉ giữ lại từ đầu tiên '''
s=input().strip().split()
res=[]
for word in s:
    if len(res)==0 or res[-1]!=word:
        res.append(word)    
print(' '.join(res))
```

```python
'''W7A25 
Tìm số có đúng k chữ số lớn nhất có thể 
trong một số có n chữ số (coi đầu vào là 
một số rất lớn - str). Không được hoán 
đổi vị trí các chữ số '''
s=input().strip()
k=int(input())
n=len(s)
res=''
for i in range(n):
    if len(res)<k:
        while len(res)+n-i>k and len(res)>0 and res[-1]<s[i]:
            res=res[:-1]
        res+=s[i]
print(res)
```

```python
'''W7A27 
Cho một dãy ngoặc ()(){}{}[][]..., kiểm tra 
xem dãy ngoặc đã cho có hợp lệ hay 
không'''
s=input().strip()
stack=[]    
for char in s:
    if char in '({[':
        stack.append(char)
    else:
        if len(stack)==0:
            stack.append(char)
            break
        top=stack.pop()
        if (top=='(' and char!=')') or (top=='{' and char!='}') or (top=='[' and char!=']'):
            stack.append(char)
            break
if len(stack)==0:
    print("YES")
else:
    print("NO")
    
```

```python
'''W7A31 
Đổi chỗ đúng 2 chữ số để được số nhỏ 
nhất có thể '''
s = input().strip()
n = len(s)
min_val = s
for i in range(n):
    for j in range(i+1, n):
        if s[i] > s[j]:
            new_s = list(s)
            new_s[i], new_s[j] = new_s[j], new_s[i]
            new_s = ''.join(new_s)
            if new_s < min_val:
                min_val = new_s
print(int(min_val))
```
