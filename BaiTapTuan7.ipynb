```python
'''W7A1 
Viết hàm tìm kiếm nhị phân 
trên một danh sách đã 
được sắp xếp tăng dần. 
Trả về chỉ số của phần tử 
hoặc -1 nếu không tìm 
thấy. mẫu hàm: def 
binary_search(arr, left, 
right, target): '''
def binary_search(arr,left,right,target):
    if left>right:
        return -1
    mid=(left+right)//2
    if arr[mid]==target:
        return mid
    elif arr[mid]>target:
        return binary_search(arr,left,mid-1,target)
    else:
        return binary_search(arr,mid+1,right,target)

arr=[int(x) for x in input().split()]
target=int(input())
result=binary_search(arr,0,len(arr)-1,target)
print(result)
```

```python
'''W7A2 
Cho danh sách arr và một 
số nguyên x, đếm xem x 
xuất hiện bao nhiêu lần 
trong danh sách. def 
count_occurrences(arr, x) 
> int: '''
def count_occurrences(arr,x):
    count=0
    for i in arr:
        if i==x:
            count+=1
    return count
arr=[int(x) for x in input().split()]
x=int(input())
result=count_occurrences(arr,x)
print(result)
```

```python
'''W7A3 
Viết hàm thực hiện thuật 
toán bubble sort, Selection 
Sort để sắp xếp danh sách 
theo thứ tự tăng dần. Sắp 
xếp nổi bọt (Bubble Sort);'''
def bubble_sort(arr):
    n=len(arr)
    for i in range(n):
        for j in range(0,n-i-1):
            if arr[j]>arr[j+1]:
                arr[j],arr[j+1]=arr[j+1],arr[j]
    return arr
arr=[int(x) for x in input().split()]
result=bubble_sort(arr)
print(result)
```

```python
'''W7A4 
Viết hàm tìm phần tử có số 
lần xuất hiện nhiều nhất 
trong danh sách. Nếu có 
nhiều phần tử cùng số lần, 
trả về phần tử xuất hiện 
trước. '''
def most_frequent(arr):
    freq={}
    for i in arr:
        if i in freq:
            freq[i]+=1
        else:
            freq[i]=1
    max_count=-1
    result=None
    for i in arr:
        if freq[i]>max_count:
            max_count=freq[i]
            result=i
    return result
arr=[int(x) for x in input().split()]
result=most_frequent(arr)
print(result)
```

```python
'''W7A5 
Cho danh sách các số 
nguyên và một số X, đếm 
tất cả các cặp (a, b) sao 
cho a + b = X. c '''
def count_pairs(arr,x):
    count=0
    seen=set()
    for i in arr:
        complement=x-i
        if complement in seen:
            count+=1
        seen.add(i)
    return count
arr=[int(x) for x in input().split()]
x=int(input())
result=count_pairs(arr,x)
print(result)
```

```python
'''W7A6 
Bạn được cung cấp một 
danh sách các danh sách 
lồng nhau, vd [[1, 2, 3], [4, 
5], [6, 7, 8, 9]], hãy làm 
phẳng danh sách '''
def flatten(nested_list):
    flat_list=[]
    for sublist in nested_list:
        for item in sublist:
            flat_list.append(item)
    return flat_list
nested_list=eval(input())
result=flatten(nested_list)
print(result)
```

```python
'''W7A7 
Cho một danh sách các số 
nguyên nums, hãy tìm độ 
dài của dãy con tăng 
nghiêm ngặt dài nhất 
(không nhất thiết phải liền 
kề). '''
def longest_increasing_subsequence(nums):
    if not nums:
        return 0
    n=len(nums)
    dp=[1]*n
    for i in range(1,n):
        for j in range(i):
            if nums[i]>nums[j]:
                dp[i]=max(dp[i],dp[j]+1)
    return max(dp)
nums=[int(x) for x in input().split()]
result=longest_increasing_subsequence(nums)
print(result)
```

```python
'''W7A9 
Cho một danh sách các số 
nums, với mỗi phần tử 
nums[i], hãy đếm xem có 
bao nhiêu phần tử bên phải 
nhỏ hơn nums[i].  
Trả về danh sách các số 
đếm. 
Efficient solution: using 
heap or binary search tree '''
def count_smaller(nums):
    result=[]
    sorted_list=[]
    for num in reversed(nums):
        left,right=0,len(sorted_list)
        while left<right:
            mid=(left+right)//2
            if sorted_list[mid]<num:
                left=mid+1
            else:
                right=mid
        result.append(left)
        sorted_list.insert(left,num)
    return result[::-1]
nums=[int(x) for x in input().split()]
result=count_smaller(nums)
print(result)
```

```python
'''W7A10 
Cho một danh sách các số 
nguyên nums và một số 
nguyên m, hãy tìm tổng 
mảng con lớn nhất theo 
modulo m.  
Tức là, với bất kỳ mảng 
con nums[i:j] nào, hãy tính 
sum(nums[i:j]) % m và trả 
về kết quả lớn nhất có thể. 
Gợi ý: 
Duy trì tổng tiền tố (prefix 
sum) mod m và sử dụng 
tìm kiếm nhị phân để tìm 
tổng tiền tố nhỏ nhất chỉ 
lớn hơn tổng hiện tại một 
chút (để tối đa hóa hiệu số 
bao quanh). '''
def max_subarray_modulo(nums,m):
    prefix_sums=[]
    current_sum=0
    max_mod_sum=0
    for num in nums:
        current_sum=(current_sum+num)%m
        max_mod_sum=max(max_mod_sum,current_sum)
        import bisect
        idx=bisect.bisect_right(prefix_sums,current_sum)
        if idx<len(prefix_sums):
            mod_sum=(current_sum - prefix_sums[idx] + m)%m
            max_mod_sum=max(max_mod_sum,mod_sum)
        bisect.insort(prefix_sums,current_sum)
    return max_mod_sum
nums=[int(x) for x in input().split()]
m=int(input())
result=max_subarray_modulo(nums,m)
print(result)
```

```python
'''W7A13 
Dãy hài hòa là một dãy 
thỏa mãn điều kiện giá trị 
lớn nhất và giá trị nhỏ nhất 
của dãy khác nhau đúng 
bằng \( 1 \). 
Viết hàm trả về độ dài của 
dãy hài hòa con dài nhất có 
thể có trong mảng giá trị 
nguyên \( arr \) có \( n \) 
phần tử 
Ví dụ: 
Input: 1 3 2 2 5 2 3 7 
Output: 5 - Dãy hài hòa con 
dài nhất là [3,2,2,2,3] '''
def findLHS(arr):
    from collections import Counter
    count=Counter(arr)
    max_length=0
    for key in count:
        if key+1 in count:
            max_length=max(max_length,count[key]+count[key+1])
    return max_length
arr=[int(x) for x in input().split()]
result=findLHS(arr)
print(result)
```
