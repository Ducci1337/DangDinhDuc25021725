```python
#W6A16 
#Nhập vào mảng n hàng, m cột số nguyên. In lại mảng ra màn hình như ma 
#trận hai chiều, định dạng mỗi phần tử 4 vị trí căn phải
n,m=map(int,input().split())
a=[]
for i in range(n):
    b=list(map(int,input().split()))
    a.append(b)

for i in range(n):
    for j in range(m):
        print('    ',a[i][j],end='')
    print()
```

```python
#W6A17 
#Nhập vào mảng vuông 2 chiều n hàng, n cột, in ra đường chéo chính và 
#đường chéo phụ của mảng
n=int(input())
a=[]
for i in range(n):
    b=list(map(int,input().split()))
    a.append(b)
for i in range(n):
    for j in range(n):
        if i==j:
            print(a[i][j],end=' ') 
print()
for i in range(n):
    for j in range(n):
        if i+j==n-1:
            print(a[i][j],end=' ')
```

```python
#W6A18 
#Nhập vào mảng n hàng, m cột số nguyên và số nguyên k (n, m, k nhập trên 
#cùng một dòng, cách nhau dấu cách). Tính và in ra tổng các phần tử trên cột 
#k (hàng k)
n,m,k=map(int,input().split())
a=[]
for i in range(n):
    b=list(map(int,input().split()))
    a.append(b)
tong=0
for i in range(n):
    tong+=a[i][k]
print(tong)
 
```

```python
#W6A19 
#Viết hàm đưa vào 1 list số nguyên, tìm và trả về vị trí có giá trị lớn nhất 
#trong list 
def vitri_max(lst):
    max_value=lst[0]
    max_index=0
    for i in range(1,len(lst)):
        if lst[i]>max_value:
            max_value=lst[i]
            max_index=i
    return max_index
```
